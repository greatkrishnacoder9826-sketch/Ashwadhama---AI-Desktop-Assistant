```python
from tkinter import *
from PIL import Image, ImageTk
import pyttsx3
import speech_recognition as sr
from deep_translator import GoogleTranslator
from tkinter.ttk import Combobox
import urllib.request
from bs4 import BeautifulSoup
from PIL import Image
import pytesseract 
from textblob import TextBlob
import subprocess


top = Tk()
top.config(background="black")
top.geometry('1500x800')

img = ImageTk.PhotoImage(file=r"C:\PYTHON NOTES\ChatGPT Image May 10, 2026, 02_05_26 AM.png")
img1 = Label(image=img)
img1.pack()

l1 = Label(top, text='Wellcome to --- Ashwadhama', font=("Arial", 20, "bold"), fg="white", bg="grey", width=800, height=2)
l1.place(x=0, y=0)

l2 = Label(top, text='Well-come to --- Ashwadhama', font=("Arial", 20, "bold"), fg="white", bg="grey")
l2.place(x=520, y=15)

l1 = Label(top, text='', font=("Arial", 20, "bold"), fg="white", bg="grey", width=20, height=800)
l1.place(x=0, y=0)

l1 = Label(top, text='', font=("Arial", 20, "bold"), fg="white", bg="black", width=20, height=1)
l1.place(x=0, y=70)


chrome_path = r"C:\Program Files\Google\Chrome\Application\chrome.exe"



def analyse_sentiment():
    text = e1.get()
    
    if text.strip() == "":
        result_label.config(text=" Pehle kuch text likho!", fg="yellow")
        return
    
    blob = TextBlob(text)
    polarity = blob.sentiment.polarity
    
    if polarity > 0:
        result_label.config(text=f" Aapka Statement POSITIVE hai!   (Score: {polarity:.2f})", fg="lime")
    elif polarity < 0:
        result_label.config(text=f" Aapka Statement NEGATIVE hai!   (Score: {polarity:.2f})", fg="red")
    else:
        result_label.config(text=f" Aapka Statement NEUTRAL hai!  (Score: {polarity:.2f})", fg="white")



def convert_text_to_audio():
    text = e2.get()
    
    if text.strip() == "":
        audio_result_label.config(text=" Pehle kuch text likho!", fg="yellow")
        return
    
    try:
        audio_result_label.config(text=" Audio bol raha hai... suno!", fg="cyan")
        top.update()
        
        engine = pyttsx3.init()
        engine.setProperty('rate', 150)
        engine.setProperty('volume', 1.0)
        engine.say(text)
        engine.runAndWait()
        
        audio_result_label.config(text=" Audio complete ho gaya!", fg="lime")
        
    except Exception as e:
        audio_result_label.config(text=f" Error: {str(e)}", fg="red")



def voice_search():
    r = sr.Recognizer()
    
    try:
        voice_result_label.config(text=" Bol kuch...", fg="cyan")
        top.update()
        
        with sr.Microphone() as source:
            audio = r.listen(source, 0, 5)  
        
        voice_result_label.config(text=" Sun liya! Samajh raha hun...", fg="yellow")
        top.update()
        
        text = r.recognize_google(audio)  
        
       
        e3.delete(0, END)
        e3.insert(0, text)
        
        voice_result_label.config(text=f" Tune ye kaha: '{text}' — Chrome mein khul raha hai!", fg="lime")
        top.update()
        
        
        url = 'https://www.google.com/search?q=' + text
        subprocess.Popen([chrome_path, url]) 
        
    except Exception as e:
        voice_result_label.config(text=f" Error: {str(e)}", fg="red")



l2 = Button(top, text="Analiyse your Sentiment", font="arial 10 bold", fg="black", bg="white", width=42, height=1, command=analyse_sentiment)
l2.place(x=0, y=120)

e1 = Entry(top, font="arial 20 bold", fg='black', bg='white', width=180)
e1.place(x=350, y=122)

result_label = Label(top, text="", font=("Arial", 14, "bold"), fg="white", bg="black")
result_label.place(x=350, y=160)



l3 = Button(top, text="Convert your Text - Audio", font="arial 10 bold", fg="black", bg="white", width=42, height=1, command=convert_text_to_audio)
l3.place(x=0, y=360)

e2 = Entry(top, font="arial 20 bold", fg='black', bg='white', width=180)
e2.place(x=350, y=360)

audio_result_label = Label(top, text="", font=("Arial", 14, "bold"), fg="white", bg="black")
audio_result_label.place(x=350, y=400)



l4 = Button(top, text="Search-by-Voice", font="arial 10 bold", fg="black", bg="white", width=42, height=1, command=voice_search)
l4.place(x=0, y=560)

e3 = Entry(top, font="arial 20 bold", fg='black', bg='white', width=180)
e3.place(x=350, y=560)

voice_result_label = Label(top, text="", font=("Arial", 14, "bold"), fg="white", bg="black")
voice_result_label.place(x=350, y=600)


top.mainloop()
```
